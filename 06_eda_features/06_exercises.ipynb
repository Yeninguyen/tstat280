#### Exercise 8

Perform a log transformation on `Salary` and then standardize the transformed values.

```python
# PUT YOUR CODE HERE FOR THE TRANSFORMATION AND STANDARDIZATION
import numpy as np

df["log_salary"] = np.log(df["Salary"])
df["salary_standardized"] = df["log_salary"].pipe(lambda x: (x-x.mean())/x.std())
df.describe()
```

Now make a visualization to view the distribution of this new transformed and standardized variable.

```python
# PUT YOUR CODE FOR THE VISUALIZATION HERE
sns.histplot(data=df, x="salary_standardized")
```
