### Practice Quiz

#### Part 1

Read in the dataset in `math_education.csv` and display the first 5 rows of the dataset.

```python
# PUT YOUR CODE HERE
import pandas as pd
df = pd.read_csv("../data/math_education.csv")
df.head()
```

#### Part 2

Drop any duplicate rows (rows where every value is the same).

```python
# PUT YOUR CODE HERE
df[df.duplicated(keep=False)]
df = df.drop_duplicates()
df.duplicated().sum()
```

#### Part 3

Treat any missing values for `study_hours` by replacing them with the median of `study_hours`.

```python
# PUT YOUR CODE HERE
df["study_hours"].isnull().sum()
df["study_hours"] = df["study_hours"].fillna(df["study_hours"].median())
df["study_hours"].isnull().sum()
```

#### Part 4

Do the following with the `time_per_problem` variable:

- Clean up the text
- Convert to `float`
- Convert to everything to seconds

```python
# PUT YOUR CODE HERE
is_minutes = df["time_per_problem"].str.contains("minutes")

df["time_per_problem"] = (
    df["time_per_problem"].str.replace("minutes", "")
    .str.replace("seconds", "")
    .str.strip()
    .astype("float")
    * is_minutes.map(lambda x: 60 if x else 1)
)

df["time_per_problem"]
```

#### Part 5

Remove any unwanted text ("%" and "percent") from the `test_score` variable and then convert this to a `float`.

```python
# PUT YOUR CODE HERE
df["test_score"] = ( df["test_score"].str.replace("%", "")
                    .str.replace("percent", "")
                    .str.strip()
                    .astype("float")
                   )
df["test_score"].head(15)
```

#### Part 6

Standardize the categories in `learning_style` using the following mapping:

- Audio, auditory, AUDITORY, Auditory → Auditory
- kinesthetic, KINESTHETIC, Kinesthetic (with space), Kinesthetic → Kinesthetic
- visual, VISUAL, Visual → Visual

(Note that one of the values for the Kinesthetic category has one extra space that needs to be removed)

```python
# PUT YOUR CODE HERE
learning_style_map = {
    "Audio": "Auditory",
    "auditory": "Auditory",
    "AUDITORY": "Auditory",
    "Auditory": "Auditory",
    "kinesthetic": "Kinesthetic",
    "KINESTHETIC": "Kinesthetic",
    "Kinesthetic ": "Kinesthetic",  
    "Kinesthetic": "Kinesthetic",
    "visual": "Visual",
    "VISUAL": "Visual",
    "Visual": "Visual",
}

df["learning_style"] = df["learning_style"].map(learning_style_map)
df["learning_style"].head(20)
```
