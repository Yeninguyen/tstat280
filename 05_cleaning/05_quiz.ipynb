### Quiz

#### Part 1

Read in the dataset in `nutrition.csv` and display the first 5 rows of the dataset.

```python
# PUT YOUR CODE HERE
import pandas as pd
df = pd.read_csv("../data/nutrition.csv")
df.head()
```

#### Part 2

Drop any duplicate rows (rows where every value is the same).

```python
# PUT YOUR CODE HERE
df[df.duplicated(keep=False)]
df.duplicated().sum()
df = df.drop_duplicates()
df.duplicated().sum()
```

#### Part 3

Treat any missing values for `calories` by replacing them with the mean of `calories`.

```python
# PUT YOUR CODE HERE
df["calories"].isnull().sum()
df["calories"] = df["calories"].fillna(df["calories"].mean())
df["calories"].isnull().sum()
```

#### Part 4

Do the following with the `protein` variable:

- Clean up the text
- Convert to `float`
- Convert to everything to ounces (1 oz is 28.3495 grams)

```python
# PUT YOUR CODE HERE
is_gram = df["protein"].str.contains("g")
df["protein"] = (
    df["protein"].str.replace("g", "")
    .str.replace("oz", "")
    .str.strip()
    .astype("float")
    *is_gram.map(lambda x:28.3495 if x else 1)
)
df["protein"]
```

#### Part 5

Remove any unwanted text ("bmi" and "BMI") from the `BMI` variable and then convert this to a `float`.

```python
# PUT YOUR CODE HERE
df["BMI"] = (
    df["BMI"].str.replace("bmi", "")
    .str.replace("BMI", "")
    .str.strip()
    .astype("float")
)
df["BMI"]
```

#### Part 6

Standardize the categories in `diet` using the following mapping:

- Omni, omnivore → Omnivore
- other, OtherDiet → Other
- vegan, VGN → Vegan
- vegetarian, Veggie → Vegetarian

```python
# PUT YOUR CODE HERE
diet_map = {
    "Omni" : "Omnivore",
    "omnivore" : "Omnivore",
    "other" : "Other",
    "OtherDiet" : "Other",
    "vegan" : "Vegan",
    "VGN" : "Vegan",
    "vegetarian" : "Vegetarian",
    "Veggie" : "Vegetarian"
}
df["diet"] = df["diet"].map(diet_map)
df["diet"]
```
